```python
import numpy as np

runs = np.array([45, 12, 78, 3, 56])
print(runs)
print(runs * 2)
```

```python
print(runs[1:3])
print(runs[:2])
print(runs[2:])
```

```python
mask = runs > 40
print(mask)
print(runs[mask])
```

```python
runs = np.array([45, 12, 78, 3, 56])
""" 
print runs which are less then 20 
guess output = [12,3]
"""
print(runs[runs < 20])
# output = [12,3]
```

```python
runs  = np.array([45, 12, 78, 3, 56])
balls = np.array([30, 10, 50, 5, 40])
```

```python
strike_rate = (runs/balls) * 100
print(strike_rate)
```

```python
print(strike_rate[strike_rate > 120])
```

```python
print(np.mean(strike_rate))
```
