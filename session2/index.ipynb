```python
import pandas as pd

df = pd.DataFrame({
    "player": ["Kohli", "Rohit", "Dhoni", "Gill", "Pant"],
    "runs":   [73, 45, 28, 91, 34],
    "balls":  [49, 38, 17, 58, 22],
    "team":   ["RCB", "MI", "CSK", "GT", "DC"],
})

print(df)
print(df.shape)       # (rows, columns)
print(df.head(3))     # pehli 3 rows
print(df["runs"])     # ek column (Series)
print(df["runs"].mean())
print(df[df["runs"] > 40])   # filter, bilkul NumPy mask jaisa
```

```python
df['strike_rate'] = (df['runs'] / df['balls'] * 100).round(2)
```

```python
df
```

```python
print(df[df['strike_rate'] > 150])
```

```python
print(df.groupby("team")["runs"].mean())
```
