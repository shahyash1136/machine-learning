```python
def predict(x, m, c):
    return m * x + c
```

```python
print(predict(10,1.2,2))
print(predict(20,1.2,2))
print(predict(30,1.2,2))
```

```python
print(predict(10, 1.2, 0))
print(predict(20, 1.2, 0))
print(predict(30, 1.2, 0))
```

```python
import numpy as np

balls = np.array([10, 20, 30])
runs = np.array([14, 22, 38])

for c in [0, 1, 2, 3]:
    error = predict(balls, 1.2, c) - runs
    total_error = np.sum(error**2)
    print(f"c={c}  errors={error}  total={total_error}")
```

```python
for c in [0, 0.25, 0.5, 0.75, 1, 1.25, 1.5]:
    error = predict(balls, 1.2, c) - runs
    total_error = np.sum(error**2)
    print(f"c={c}  errors={error}  total={total_error}")
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

balls = np.array([10, 20, 30]).reshape(-1, 1)
runs = np.array([14, 22, 38])

model = LinearRegression()
model.fit(balls, runs)  # seekhne wala step

print(model.coef_)  # ye m hai
print(model.intercept_)  # ye c hai
print(model.predict([[25]]))  # 25 balls par runs predict
```
